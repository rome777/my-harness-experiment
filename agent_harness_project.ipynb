하네스는 모델이 일을 하는 동안 요청, 도구, 기록, 승인, 관찰(observation)을 연결하는 실행 환경입니다. 위 그림처럼 요청이 들어오면 상태, 모델, 도구, 관찰을 거쳐 다시 상태로 돌아오는 에이전트 루프 전체를 하네스가 감쌉니다. 같은 모델도 행동을 적는 형식, 남겨 둘 컨텍스트, 허용된 권한이 달라지면 다른 결과를 냅니다. 그래서 결과의 차이를 모델의 답변 능력만으로는 설명할 수 없습니다.

### 맡긴 일이 막히는 곳을 찾는 일곱 기능

이 노드는 하네스를 일곱 기능으로 나눠 봅니다. 이 일곱 기능의 분류를 기능 지도라고 부릅니다. 기능 지도는 맡긴 일이 멈춘 이유를 찾는 데 씁니다. 셋째 절부터 아홉째 절까지 아래 순서대로 한 기능씩 다룹니다.

| 기능 | 하는 일 |
|---|---|
| 행동 포맷 | 모델이 읽고, 고치고, 실행하는 행동을 실패하기 어려운 형식으로 적게 합니다. |
| 컨텍스트 관리 | 긴 작업에서 무엇을 남기고 무엇을 줄일지 정합니다. |
| 경험·스킬 축적 | 한 번 해결한 일을 다음 작업에 다시 씁니다. |
| 성찰(reflection) | 실행 결과를 살펴 다음 행동을 고칩니다. |
| 미니멀(단순성) | 필요한 구성만 남겨 운영 부담을 줄입니다. |
| 권한·보안 | 읽기, 수정, 실행의 허용 범위와 승인 방식을 정합니다. |
| 평가 루프 | 결과가 맞았는지 어떤 근거로 확인할지 정합니다. |

기능 지도는 다음 세 질문을 차례로 던지며 씁니다.

1. 이 일은 어디서 막히는가: 행동이 애매한가, 컨텍스트가 넘치는가, 경험이 재사용되지 않는가, 승인이 많은가, 결과를 증명하기 어려운가
2. 그 병목을 줄일 기능은 무엇인가
3. 그 기능이 실제로 작동했다는 기록은 어디에서 확인할 수 있는가

이 세 질문으로 제품 설명을 보기 전에 맡길 일의 범위를 먼저 정합니다. 일곱 도구 비교표 절에서 도구 후보를 줄일 때도 이 세 질문을 다시 씁니다.

### 기능 지도와 비교 카드의 차이

기능 지도의 일곱 기능은 맡길 일의 병목을 진단하는 분류입니다. 비교 카드는 도구 화면과 문서를 같은 방식으로 관찰하기 위한 여섯 칸짜리 기록 양식입니다. 기능 지도에서 컨텍스트 관리가 중요하다고 판단한 뒤에도, 카드에서는 그 도구의 상태와 주요 동작이 실제로 어떻게 보이는지 확인합니다. 이후 절에서는 주요 도구를 이 여섯 칸으로 소개합니다. 한 기능만 살펴보는 도구(Hermes, 평가 루프 절의 SWE-agent)는 그 기능에 해당하는 항목만 표로 정리합니다.

| 카드 칸 | 관찰할 내용 |
|---|---|
| 진입점 | 요청이 들어가는 화면, 명령, 연결 지점을 확인합니다. |
| 상태 | 세션, 대화, 작업 기록이 남는 위치를 확인합니다. |
| 주요 동작 | 읽기, 검색, 수정, 실행 가운데 화면에서 확인한 행동을 적습니다. |
| 권한 경계 | 읽기 밖의 행동 전에 승인이나 제한이 보이는지 확인합니다. |
| 1차 자료 | 실행 기록, 설정, 공식 문서처럼 나중에 다시 열 수 있는 근거를 연결합니다. |
| 한계 | 이번 관찰만으로 확인하지 못한 범위를 구분합니다. |

## 코딩 에이전트 화면에서 하네스 신호 관찰하기

제품 설명을 읽는 데서 그치지 않고, 지금 쓰는 도구가 요청을 받은 순간부터 화면에 무엇을 보여 주는지 직접 관찰합니다. 관찰할 도구는 다음 둘 가운데 이미 쓰고 있는 하나입니다.

| 도구 | 설명 |
|---|---|
| [Claude Code](https://code.claude.com/docs/en/overview) | Anthropic이 만든 터미널용 코딩 에이전트입니다. 터미널 안에서 저장소를 읽고, 파일을 고치고, 명령을 실행합니다. |
| [opencode](https://opencode.ai/docs/) | 오픈소스 터미널 코딩 에이전트입니다. 여러 모델 제공자를 설정에서 골라 쓸 수 있습니다. |

수업 환경에는 두 도구가 설치되어 있습니다. 개인 컴퓨터에서 실습하려면 아래 "설치 안내"를 따라 하나를 먼저 설치합니다. 이번 실습에서는 로그인 변경, 확장 연결, 파일 수정을 하지 않습니다. 요청은 읽기 전용 작업 하나면 충분합니다. 현재 작업 폴더의 파일 목록을 읽는 요청이나 공개 문서 한 건을 요약하는 요청이 알맞습니다. 이번 실습은 결과의 좋고 나쁨을 평가하지 않고, 요청이 들어가는 곳과 실행 범위가 화면에 어떻게 보이는지 찾습니다.

### 관찰 범위 정하기

작업 폴더는 기존 작업 자료와 분리하고, 그 안에 간단한 fixture 하나만 둡니다. fixture는 시험할 때마다 결과가 똑같이 나오도록 미리 준비해 둔 고정 자료를 가리키는 테스트 용어입니다. 도구의 행동을 보려면 자료가 늘 같아야, 달라진 것이 자료가 아니라 도구라고 말할 수 있습니다.

> **비유로 이해하기**
> fixture는 요리 교실에서 매번 같은 재료를 미리 다져 두는 것과 비슷합니다. 재료(fixture)가 그때그때 다르면 결과의 차이가 요리사(도구)의 실력 때문인지 재료 때문인지 구분할 수 없습니다.
> 다른 점: 요리 재료는 매번 새로 준비하지만, fixture는 같은 파일을 그대로 다시 씁니다.

이 관찰에서는 공개 문서 한 건 또는 내용이 알려진 작은 텍스트 파일 한 건이면 충분합니다. 비밀값, 개인 자료, 실제 저장소, 수정이 필요한 파일은 넣지 않습니다.

관찰 메모에는 도구 이름과 버전, 작업 폴더 위치, fixture 종류를 먼저 적습니다. 나중에 같은 화면을 다시 볼 때 어떤 조건에서 본 것인지 구분하기 위해서입니다.

| 기록 위치 | 현재 대화, 세션 목록, 작업 기록, 로그 가운데 실행 결과가 남는 곳 | 경험·스킬 축적, 성찰 |
| 승인 신호 | 읽기 밖의 행동 전에 보이는 승인 창, 허용 선택지, 실행 예정 명령, 권한 설정 안내 | 권한·보안 |

- 입력과 도구 호출은 행동 포맷이 화면에서 어떤 순서와 단위로 드러나는지 보여 줍니다. 요청 문장만 보이고 행동이 보이지 않으면, 주요 동작은 확인하지 못한 것으로 적습니다.
- 관찰 축약과 기록 위치는 화면에서 실제로 보이는 위치만 적습니다.
- 승인 신호가 보이면 그 문구와 위치를 적습니다. 아무 신호도 보이지 않으면 이번 읽기 요청에서는 승인 신호를 확인하지 못했다고 적습니다.
- 이번 실습에서 확장 기능을 열거나 추가하지는 않습니다. 화면에 도구 연결, 플러그인, MCP, 설정 파일 같은 단서가 보이면 관찰한 사실과 확인하지 못한 범위를 나눠 적습니다.
- 미니멀이 필요한지, 평가 루프가 어디에서 이루어지는지는 이 짧은 관찰로 결론 내리지 않습니다.

### 시범: Claude Code에서 파일 목록 요청 관찰하기

아래는 macOS의 Claude Code CLI 2.1.220에서 새 작업 폴더 `~/obs-lab`을 열고 실제로 실행한 기록입니다. 폴더에는 공개 문서 한 건을 받아 둔 `notes.txt`만 있습니다. 버전 번호는 화면에 표시된 값을 그대로 관찰 조건으로 적었습니다. Windows에서는 경로와 출력 형식이 다르게 보일 수 있으니, 실습에서는 자기 화면에 보인 그대로 적습니다.

**요청 입력.** 대화창에 요청을 입력합니다.

```
이 폴더의 파일 목록을 보여 줘
```

요청 문장은 대화 흐름 안에 남고, 응답 본문과 구분되어 표시됩니다. 결과 목록의 내용보다 요청이 들어간 화면의 위치를 먼저 봅니다.

**도구 호출 표시.** 요청을 넣으면 응답이 만들어지는 동안 도구 호출이 화면에 표시됩니다. 이 실행에서 Claude Code는 파일 목록을 Bash로 조회했습니다.

```
● Bash(ls -la /Users/admin/obs-lab)
    List files in obs-lab
```

Claude Code의 [도구 목록 문서](https://code.claude.com/docs/en/tools-reference)에는 파일 탐색용 `Glob`과 파일 읽기용 `Read`가 있지만, 이 요청에서 모델은 Bash의 `ls -la`를 골랐습니다. 화면에 나타난 도구 이름과 명령이 이번 실행의 사실입니다. 어떤 도구를 쓸지는 모델이 판단했습니다. 버전과 운영체제에 따라 기본으로 켜진 도구가 다르므로, 실습에서는 자기 화면에 나온 도구 이름을 그대로 적습니다.

**도구 결과.** 도구의 결과가 돌아옵니다.

```
  total 8
  drwxr-xr-x   4 admin  staff   128 Sep  2 01:36 .
  drwxr-xr-x@ 171 admin  staff  5472 Sep  2 01:38 ..
  drwxr-xr-x   4 admin  staff    128 Sep  2 01:37 .omc
  -rw-r--r--   1 admin  staff    32 Sep  2 01:36 notes.txt
```

명령의 출력이 원본 그대로 돌아오고, 이 원본이 세션 기록에 남습니다. 컨텍스트 관리는 이렇게 쌓이는 기록을 다룹니다. 지금은 네 줄이라 그대로 두었지만, 출력이 길어지면 관찰 축약이 적용될 수 있습니다. `.omc`는 시범 환경에만 있는 상태 폴더이므로 여러분의 폴더에는 없을 수 있습니다.

**최종 응답.** 최종 응답이 완성됩니다.

```
  `/Users/admin/obs-lab` 내용:

  - `notes.txt` — 32바이트 파일
  - `.omc/` — OMC 상태 디렉터리 (숨김)
```

모델은 도구 결과를 읽고 사람이 읽을 요약을 만들었습니다. 요청, 도구 호출, 도구 결과, 요약이 한 세션에 순서대로 남았습니다. 이 실행에서 화면에 남은 기록은 이 네 가지입니다.

**승인 창.** 이 실행에서 승인 창은 나타나지 않았습니다. Claude Code는 `ls`, `cat`, `grep` 같은 읽기 전용 명령을 승인 없이 실행합니다. (`[공식 문서]` Anthropic, [Claude Code 권한 문서](https://code.claude.com/docs/en/permissions)) 승인 창이 없었다는 사실은 적되, 설정 파일에 어떤 허용 규칙이 있는지는 열어 보지 않았으므로 미확인으로 둡니다. 관찰 축약 표시와 세션 기록 위치도 이 화면에서 보인 범위만 적습니다.

**시범의 관찰 메모**

> 조건: 도구는 Claude Code CLI 2.1.220, 작업 폴더는 `~/obs-lab`, fixture는 공개 문서를 받아 둔 `notes.txt` 한 건.
> 입력: 요청은 대화창에 들어가고 응답과 구분되어 남았습니다.
> 도구 호출: Bash(`ls -la`)가 화면에 표시되었습니다. 도구 목록에 있는 Glob/Read가 아니라 모델이 Bash를 골랐습니다.
> 관찰: 원본 `ls` 출력 네 줄이 그대로 돌아왔습니다. 축약 표시는 이 실행에서 확인하지 못했습니다.
> 승인: 이 읽기 요청에는 승인 창이 없었습니다. 설정의 허용 규칙은 미확인입니다.
> 기록: 세션에 요청부터 요약까지 순서대로 남았습니다.

이 메모에는 "이 도구는 읽기가 안전하다"는 해석을 쓰지 않습니다. 안전하다고 판정하려면 권한 규칙을 확인해야 하는데, 이번 관찰에서는 그 규칙을 열어 보지 않았기 때문입니다.

### 실습: 내 도구에서 관찰하기

**1단계. 작업 폴더와 fixture 준비**

기존 작업 자료와 분리된 새 폴더를 만듭니다. Windows의 PowerShell이나 명령 프롬프트, macOS·Linux의 터미널에서 같은 명령을 씁니다.

```bash
mkdir obs-lab
cd obs-lab
```

폴더 안에 `notes.txt`를 만들고 공개 문서 몇 줄을 붙여 넣어 저장합니다. 이 파일이 이번 관찰의 fixture입니다.

**2단계. 관찰 조건 적기**

메모 맨 위에 도구 이름과 버전, 작업 폴더 위치, fixture 종류를 적습니다. 버전은 화면에 표시된 값을 그대로 적습니다. Claude Code는 터미널에서 `claude --version`으로도 버전을 확인할 수 있습니다.

**3단계. 도구를 열고 읽기 요청 보내기**

`obs-lab` 폴더에서 도구를 실행합니다. opencode를 쓰면 `claude` 대신 `opencode`를 실행합니다.

```bash
claude
```

대화창에 시범과 같은 요청을 입력합니다.

```
이 폴더의 파일 목록을 보여 줘
```

**4단계. 다섯 신호를 찾아 관찰 메모 쓰기**

아래 틀에 자기 화면에서 본 것만 채웁니다. 보이지 않은 신호는 "이번 실행에서 확인하지 못함"으로 적습니다.

```
조건: 도구 ___ (버전 ___), 작업 폴더 ___, fixture ___
입력:
도구 호출:
관찰: (축약 표시가 보였는가)
승인: (승인 창이 보였는가, 설정의 허용 규칙은 확인했는가)
기록: (요청부터 요약까지 어디에 남았는가)
```

Claude Code에서는 `/permissions`로 현재 허용 규칙을 볼 수 있습니다. 승인 칸을 미확인으로 남겼다면 나중에 이 명령으로 확인합니다.

### 사실과 해석 나눠 적기

기록에는 화면이나 로그에서 확인한 내용만 넣습니다. 화면에 보인 연결이나 설정 단서는 확인한 사실로, 이번 실행에서 보이지 않은 기능은 한계로 구분합니다.

같은 내용도 적는 방식에 따라 사실이 되기도 하고 해석이 되기도 합니다. 파일 목록을 요청했더니 목록이 화면에 나오고 승인 창은 뜨지 않았다면 다음처럼 한 줄로 적습니다.

> 파일 목록 요청에 승인 없이 읽기가 실행되었습니다. 관찰한 사실: 화면의 파일 목록. 미확인: 설정 파일의 권한 규칙

이 기록은 도구 전체를 판정하는 근거가 아니라, 짧은 실행 하나에서 보인 범위와 보이지 않은 범위를 나눈 것입니다. 해석에 필요한 근거는 뒤의 기능별 절에서 모읍니다.

## 행동 포맷: 편집과 실행을 적는 형식

모델이 고칠 내용을 알고 있어도 변경 위치와 형식을 잘못 적으면 작업은 실패합니다. 행동 포맷은 모델이 바꿀 대상, 바꿀 내용, 실행 결과를 서로 확인할 수 있는 단위로 적게 해서 이런 실패를 줄이는 기능입니다.

자유 문장으로 "이 함수를 고쳐 달라"고 하면 모델은 어디까지가 대상인지, 기존 내용과 교체 내용이 무엇인지를 다시 추정해야 합니다. 찾을 내용과 바꿀 내용을 묶은 편집 블록이나 실행 가능한 코드는 다음 행동의 형태를 좁힙니다. Claude Code와 opencode에서 파일 변경과 명령 실행이 어떤 표시로 남는지 살피는 이유도 여기에 있습니다. 권한과 컨텍스트의 범위는 뒤 절에서 따로 다룹니다.

### 행동 포맷을 비교한 연구와 도구

| 연구·도구 | 비교한 것 | 결과 (해당 조건) | 근거 |
|---|---|---|---|
| SWE-agent | 전용 인터페이스(ACI)에서 편집 기능을 뺀 조건 | 해결률 18.0%에서 10.3%로 하락 (SWE-bench Lite, GPT-4 Turbo) | `[학술]` |
| CodeAct | 행동을 JSON 도구 호출 대신 실행 가능한 Python 코드로 쓰게 함 | 17개 LLM에서 기존 형식보다 성공률이 최대 20% 높음 | `[학술]` |
| Aider | 파일 전체를 다시 쓰는 `whole` 포맷과 바꿀 부분만 쓰는 `diff` 포맷 | gpt-3.5-turbo-0301 첫 시도 통과율: `whole` 46%, `diff` 30% | `[벤더]` |

SWE-bench Lite는 실제 GitHub 이슈를 풀게 하는 벤치마크인 SWE-bench(평가 루프 절에서 다룸)에서 300문제를 골라 만든 축소판입니다. ablation은 구성요소를 하나씩 빼거나 바꿔 성능 차이를 보는 실험입니다.

**SWE-agent**

[SWE-agent](https://github.com/SWE-agent/SWE-agent)는 프린스턴 연구진이 만든 연구용 코딩 에이전트입니다. 언어 모델이 쓰기 좋게 만든 전용 인터페이스를 ACI(Agent-Computer Interface)라고 부르고, ACI가 일반 셸보다 적합할 수 있다는 가설을 검토했습니다. SWE-bench Lite ablation에서 기본 조건의 해결률은 18.0%였고, 편집 기능을 뺀 조건은 10.3%였습니다. 이 차이로 인터페이스 설계가 해결률을 바꿀 수 있다는 것까지는 말할 수 있습니다. 같은 벤치마크와 구현 조건에서 나온 결과이므로, 다른 모델과 편집 도구에서도 같은 크기의 차이가 난다고 보지는 않습니다. (`[학술]` Yang et al., NeurIPS 2024, [arXiv:2405.15793](https://arxiv.org/abs/2405.15793))

**CodeAct**

[CodeAct](https://arxiv.org/abs/2402.01030)는 행동을 실행 가능한 Python 코드 한 덩어리로 쓰게 하는 방식을 제안한 연구입니다. 여러 파일의 줄 수를 세는 작업을 예로 들면 두 방식은 다음처럼 다릅니다.

- 도구 호출 방식: 파일 열기를 호출하고, 결과를 받고, 다시 줄 세기를 호출하는 식으로 여러 번 오갑니다.
- 코드 방식: for 루프와 함수 호출이 든 코드 한 덩어리를 쓰고 실행 결과만 받습니다. 결과를 보고 코드를 고쳐 다시 제출할 수도 있습니다.

흩어진 도구 호출을 코드 하나로 묶으면 모델이 계획 전체를 한 번에 표현할 수 있다는 것이 이 연구의 가설입니다. 17개 LLM에서 기존 형식보다 성공률이 최대 20% 높았다고 보고했습니다. (`[학술]` Wang et al., ICML 2024, [arXiv:2402.01030](https://arxiv.org/abs/2402.01030))

**Aider**

[Aider](https://aider.chat/)는 터미널 안에서 파일을 편집하는 오픈소스 도구로, 변경을 자동으로 git 커밋합니다. 편집 포맷은 두 가지입니다.

| 포맷 | 모델이 출력하는 것 | `hello.py`의 한 줄을 바꿀 때 |
|---|---|---|
| `whole` | 파일 전체를 새 내용으로 다시 씁니다. 바뀌지 않은 줄도 모두 출력합니다. | 파일의 모든 줄을 다시 씁니다. |
| `diff` | 바꿀 부분만 search-replace 블록으로 씁니다. | 바꾸기 전 줄과 바꾼 뒤 줄만 짝으로 적습니다. |

Aider의 수치는 독립 비교가 아니라 Aider가 직접 만든 벤치마크의 결과입니다. 측정 방식은 다음과 같습니다.

- 133개 Exercism Python 작업을 풀게 합니다.
- 실패하면 테스트 오류의 앞 50줄을 다시 넣어 재시도하게 합니다.
- 성공 기준은 전체 테스트 통과입니다.

실패했을 때 개발자가 보는 것과 같은 테스트 오류를 돌려준다는 점에서 실제 개발 흐름과 비슷한 설계입니다. 작성한 코드가 실제로 동작하는지 잰다는 점은 SWE-agent와 같지만, 작업 난이도와 채점 방식이 다르므로 두 벤치마크의 수치를 나란히 놓지 않습니다. 같은 모델에서도 포맷에 따라 통과율이 갈렸지만, 다른 작업이나 모델에서도 `whole`이 `diff`보다 낫다는 뜻은 아닙니다. (`[벤더]` Aider, [자체 벤치마크](https://aider.chat/docs/benchmarks.html))

### Aider 비교 카드

- **진입점:** 터미널 대화에서 파일과 요청을 함께 받아 편집 대화를 시작합니다.
- **상태:** 대화, 지정 파일, 테스트 결과가 다음 응답에 쓰입니다.
- **주요 동작:** `search`와 `replace`를 한 블록에 써서 기존 문자열과 교체 문자열을 함께 나타냅니다.
- **권한 경계:** 파일 변경과 명령 실행의 허용 범위는 설치 버전, 설정, 실제 실행 기록에서 다시 확인해야 합니다.
- **1차 자료:** 공개 문서의 편집 포맷 설명과 자체 벤치마크입니다. 독립 재현 결과는 아닙니다.
- **한계:** 검색 문자열이 파일 내용과 그대로 일치하지 않거나 컨텍스트가 오래되면 구조화된 블록도 의도와 다른 변경을 낼 수 있습니다.

도식은 편집 블록이 변경 의도를 드러내는 흐름을 보여 줍니다. 블록이 변경의 정확성이나 실행 안전을 보장하지는 않으므로, 대상 파일, 일치 결과, 테스트 결과, 승인 표시를 함께 봅니다.

### SWE-agent 비교 카드

- **진입점:** 이슈 설명과 저장소 환경이 ACI를 거쳐 모델의 작업 요청으로 들어갑니다.
- **상태:** 검색, 파일 열람, 편집, 실행 관찰이 다음 판단에 필요한 형식으로 남습니다.
- **주요 동작:** 구조화된 `edit` 명령과 lint 피드백이 자유 셸 편집보다 좁은 편집 단위를 제공합니다.
- **권한 경계:** 논문의 ACI는 성능 실험용 인터페이스이며, 실제 환경의 파일·명령 권한 정책을 보여 주지는 않습니다.
- **1차 자료:** NeurIPS 2024 논문이 ACI 구성과 SWE-bench Lite ablation 결과를 공개합니다.
- **한계:** 18.0%와 10.3%는 편집 기능을 뺀 해당 조건의 결과이며, 모델 세대와 작업이 바뀌면 차이도 달라질 수 있습니다.

### 포맷이 결과를 좌우하지 않는 경우

Aider 벤치마크에서 GPT-4는 `whole`과 `diff`의 결과가 거의 같았고, Aider는 비용 때문에 `diff`를 유지했다고 설명합니다. 포맷이 언제나 결과를 정한다고 말할 수 없습니다. 다음 조건에서는 포맷보다 다른 실패 원인이 더 클 수 있습니다.

- 모델이 어느 포맷으로도 수정 내용을 틀리지 않고 적는 경우
- 변경 범위가 아주 작은 경우
- 검색 문자열이 불안정한 경우

### 비교 실험: 포맷만 바꾸기

| 구분 | 내용 |
|---|---|
| 고정 | 작은 fixture 하나와 테스트 하나 |
| 바꾸는 것 | 같은 수정을 자유 텍스트로 요청하는 조건과 `search`·`replace` 구조 포맷으로 요청하는 조건 |
| 기록 | 대상 파일을 처음 맞힌 횟수, 변경 뒤 테스트 통과 여부, 재시도 횟수 |

한 번의 결과로 일반 결론을 내지 않고, 모델, 작업, 포맷 조건을 함께 적습니다. 위 실험은 요청 문장의 형식을 바꾸는 것이고, 하네스가 모델에게 요구하는 편집 포맷 자체를 바꾸려면 Aider처럼 포맷을 고를 수 있는 도구가 필요합니다. Aider는 `--edit-format` 옵션으로 `whole`, `diff` 같은 포맷을 지정합니다. (`[공식 문서]` Aider, [편집 포맷 문서](https://aider.chat/docs/more/edit-formats.html))

## 컨텍스트 관리: 남길 정보와 다시 찾을 정보

컨텍스트는 모델이 한 번의 요청에서 입력으로 읽는 내용 전체이고, 한 번에 넣을 수 있는 최대 크기를 컨텍스트 윈도우(context window)라고 합니다. 컨텍스트 윈도우가 크면 한 번에 읽을 수 있는 기록은 늘어나지만, 컨텍스트 윈도우가 넓다고 지금 작업에 필요한 정보가 눈에 잘 띄는 것은 아닙니다.

긴 실행에서는 명령 결과, 오류 메시지, 파일 내용, 중간 판단이 계속 쌓입니다. 처음에는 유용했던 단서도 새 관찰이 이어지면 대화 위쪽으로 밀려나 찾기 어려워집니다. 컨텍스트 관리는 다음 행동에 필요한 관찰과 저장소 구조를 골라 다시 보여 주어, 중요한 실패 조건이나 다음에 열 파일의 단서가 긴 이력에 묻히지 않게 하는 기능입니다.

컨텍스트 관리는 정해진 토큰 예산 안에 무엇을 넣고, 지금 빠진 정보는 어디에서 다시 찾을지를 정하는 일입니다. 같은 크기의 컨텍스트 윈도우라도 무엇으로 채우는지에 따라 다음 행동이 달라집니다.

### 컨텍스트를 고르는 방법

| 방법 | 하는 일 | 근거 |
|---|---|---|
| 최근 관찰 유지와 과거 관찰 축약 (SWE-agent) | 최근 관찰 5개는 자세히 남기고 그 전 관찰은 줄여서 남깁니다. | `[학술]` |
| repo map (Aider) | 파일 전체 대신 저장소의 뼈대만 담은 목록을 넣습니다. | `[벤더]` 설계 글, 독립 성능 비교 아님 |
| 반복 검색 (RepoCoder) | 모델이 만든 코드로 다시 검색해 더 관련 있는 코드 조각을 찾습니다. | `[학술]` |
| 검색해 다시 불러오기 | 요약하거나 뺀 정보를 필요할 때 다시 검색해 컨텍스트에 넣습니다. | 근거 없음 (아래 비교 실험으로 확인) |
| 서브에이전트의 격리 컨텍스트 | 서브에이전트가 좁은 컨텍스트에서 따로 일하고 결과만 돌려줍니다. | `[공식 문서]` 구조 설명 (아래 비교 카드), 효과 비교 아님 |

이 방법들은 무조건 줄이는 것이 목적이 아니라, 비용과 정확도를 함께 따져 넣을 양을 정하는 방법입니다.

**최근 관찰 유지 (SWE-agent)**

SWE-agent 연구진은 SWE-bench Lite에서 최근 관찰 다섯 개는 자세히 남기고 과거 관찰은 축약한 조건을 전체 이력을 남긴 조건과 비교했습니다. 최근 다섯 개 관찰 조건은 18.0%, 전체 이력 조건은 15.0%로 3.0%p 높았습니다. 같은 작업과 모델에서 관찰을 고르는 방식만 바꿔도 결과가 달라졌다는 뜻입니다. 다섯 개가 모든 작업에서 가장 좋다는 뜻은 아닙니다. (`[학술]` Yang et al., NeurIPS 2024, [arXiv:2405.15793](https://arxiv.org/abs/2405.15793))

**repo map (Aider)**

큰 저장소의 파일을 전부 넣으면 컨텍스트 윈도우가 곧 넘칩니다. repo map은 저장소의 뼈대만 담은 목록으로, 어떤 클래스와 함수가 어느 파일에 정의되어 있는지 보여 줍니다. Aider는 파일 사이의 정의·참조 관계로 그래프를 만들어 파일의 순위를 매기고, 기본 1천 토큰 예산 안에서 넣을 항목을 고릅니다. 모델은 파일 전체가 아니라 이 목록을 보고 필요한 파일만 편집 요청에 첨부합니다. (`[벤더]` Aider, [설계 글](https://aider.chat/docs/repomap.html), 2023-10-22)

**반복 검색 (RepoCoder)**

RepoCoder는 저장소에서 검색한 코드 조각을 넣어 코드를 생성하고, 생성한 코드로 다시 검색해 더 관련 있는 조각을 찾는 반복 방식을 제안했습니다. (`[학술]` Zhang et al., EMNLP 2023, [arXiv:2303.12570](https://arxiv.org/abs/2303.12570))

### Claude Code와 opencode 비교 카드 (컨텍스트 관리)

| 카드 칸 | Claude Code | opencode |
| --- | --- | --- |
| 진입점 | 메인 대화가 터미널의 작업 요청을 받습니다. | 현재 에이전트가 대화의 작업 요청을 받습니다. |
| 상태 | 서브에이전트는 맡은 좁은 컨텍스트에서 작업하고, 결과만 메인 대화에 돌려줍니다. | 에이전트별 대화 컨텍스트와 실행 단계가 분리되어 기록됩니다. |
| 주요 동작 | 조사를 맡긴 서브에이전트의 결과를 받아 메인 대화가 다음 행동을 정합니다. | 에이전트별 `steps` 상한을 두어 한 대화가 끝없이 관찰을 쌓지 않게 합니다. `steps`는 에이전트가 텍스트로만 답하도록 강제되기 전까지 반복할 수 있는 최대 횟수입니다. |
| 권한 경계 | 서브에이전트가 쓸 도구와 승인 동작은 해당 설정과 실행 기록에서 확인해야 합니다. | 에이전트별 도구 설정과 단계 상한은 설정 파일과 실행 기록에서 확인해야 합니다. |
| 1차 자료 | 공식 서브에이전트 문서, 설정 파일, 실행 기록을 함께 확인합니다. 문서는 설치 버전과 실제 권한 설정을 대신하지 않습니다. | 공식 [에이전트 설정 문서](https://opencode.ai/docs/agents), 에이전트 설정 파일, 실행 기록을 함께 확인합니다. 설정 항목은 버전에 따라 다를 수 있습니다. |
| 한계 | 격리된 작업 결과가 메인 대화에 돌아오므로, 돌아온 결과가 충분한지는 작업마다 검토해야 합니다. | 단계 상한은 관련 컨텍스트를 골라 넣는 규칙이 아니므로, 검색과 요약의 품질은 따로 확인해야 합니다. |

### 줄이는 것이 늘 답은 아닌 경우

검색이나 요약 뒤 다시 불러오는 방식이 항상 정확도를 높이지는 않습니다. RAG(검색 증강 생성)는 필요한 문서 조각을 검색해 컨텍스트에 넣는 방식이고, long-context는 긴 원문을 그대로 넣는 방식입니다. 두 방식을 비교한 연구는 예산이 충분할 때 long-context가 평균 성능에서 일관되게 RAG보다 높았고, RAG의 장점은 낮은 비용이라고 보고했습니다. (`[학술]` Li et al., EMNLP 2024 Industry Track, [arXiv:2407.16833](https://arxiv.org/abs/2407.16833))

그러므로 원문이 컨텍스트 윈도우에 다 들어가는데도 비용 때문에 줄이려는 경우라면, 정보를 줄이기 전에 정확도가 얼마나 떨어지는지 먼저 따져 봅니다. 예산이 충분하고 원문 전체가 필요한 작업인지, 비용을 낮추려고 고르고 압축할 작업인지를 함께 판단합니다.

Anthropic도 컨텍스트를 단순한 저장 공간이 아니라 요청마다 제한된 입력 예산을 나눠 쓰는 대상으로 설명합니다. (`[벤더]` Anthropic Engineering, [Effective context engineering for AI agents](https://www.anthropic.com/engineering/effective-context-engineering-for-ai-agents)) 이 관점에서는 실행 기록으로 다음을 확인합니다.

- 요약이 필요한 단서를 버리지 않았는가
- repo map이 필요한 파일을 놓치지 않았는가
- 서브에이전트의 결과가 메인 대화에 충분히 돌아왔는가

### 비교 실험: 전체 이력과 요약 뒤 다시 불러오기

같은 모델과 같은 작업에서, 전체 이력을 남긴 조건과 요약한 뒤 필요할 때 다시 검색해 넣는 조건을 비교합니다. 나머지 조건은 같게 두고 성공 여부, 사용 토큰, 다시 찾은 정보가 원문과 맞는지를 기록하면, 어느 조건에서 요약이 도움이 되고 어느 조건에서 원문이 필요한지 확인할 수 있습니다.

## 경험·스킬 축적: 지난 실행을 다음 실행에 쓰기

로그를 저장했다고 경험이 쌓인 것은 아닙니다. 다시 찾아 적용할 수 있는 형태여야 다음 실행에 쓸 수 있습니다.

작업 전체를 보관하는 일과 다음 작업에서 쓸 판단을 남기는 일은 다릅니다. 실패한 명령, 우연히 통과한 설정, 현재 작업에만 맞는 해결책까지 다시 불러오면 기록이 오히려 판단을 흐릴 수 있습니다. 경험이 다음 실행에 쓰이려면 다음 세 조건을 갖춰야 합니다.

- 원문 기록에서 근거를 확인할 수 있습니다.
- 필요할 때 검색할 수 있습니다.
- 다음 행동으로 바꿔 적용할 수 있습니다.

### 경험을 다음 행동에 쓰는 연구

| 연구 | 남기는 것 | 다음 작업에서 하는 일 | 결과 (해당 환경) |
|---|---|---|---|
| ExpeL | 여러 경험에서 뽑은 판단(insight)과 성공 사례 | 비슷한 성공 사례를 검색해 함께 제공 | WebShop 평균 보상 0.701 (ReAct 0.665) |
| Voyager | 실행 가능한 코드 스킬 라이브러리 | 필요한 스킬을 검색해 검증한 코드를 조합 | 마인크래프트에서 고유 아이템 3.3배, 이동 거리 2.3배 |
| AWM | 과거 궤적에서 뽑은 작업 절차(workflow) | 후보 workflow를 골라 현재 작업에 맞춰 사용 | Mind2Web 24.6%, WebArena 51.1% 상대 개선 |
| WikiSkill | 원문 궤적, 위키, 실행 스킬 세 층 | 세 층을 구분해 제공하고 위키로 스킬을 갱신 | 다섯 모델 모두 스킬 없는 조건보다 평균 정확도 상승 |

**ExpeL**

ExpeL은 경험에서 배우는 에이전트를 연구한 논문입니다. 한 번의 실행에서 나온 관찰과 성공·실패 경험을 그대로 다음 프롬프트에 쌓지 않고, 여러 경험에서 반복해 쓸 판단을 insight로 뽑아 저장합니다. 새 작업이 오면 벡터 검색 라이브러리 FAISS에서 kNN(가장 가까운 k개 찾기)으로 비슷한 성공 사례를 찾아 실행에 함께 넣습니다. 넣는 개수는 환경마다 다릅니다(WebShop 2개, HotpotQA 6개).

GPT-4로 insight를 뽑고 GPT-3.5로 실행한 WebShop 조건에서 평균 보상은 0.701로, 기본 에이전트 방식인 ReAct(추론과 행동을 번갈아 하는 방식)의 0.665보다 높았습니다. insight 목록은 모델이 관리합니다. 모델이 새 insight를 추가하거나 기존 insight를 고치고 찬성·반대 표를 매기며, 반대가 쌓여 중요도 점수가 0이 된 insight는 지웁니다. 다만 무엇이 틀렸는지도 모델이 판단하므로, 틀린 insight가 걸러진다는 보장은 없습니다. (`[학술]` Zhao et al., AAAI 2024, [arXiv:2308.10144](https://arxiv.org/abs/2308.10144))

**Voyager**

Voyager는 마인크래프트 세계에서 스킬을 쌓는 에이전트를 연구한 논문입니다. 자연어 메모 대신 실행 가능한 코드 스킬을 라이브러리에 남기고, 새 작업에서는 필요한 스킬을 검색해 이미 검증한 코드를 조합합니다. 기존 방법보다 고유 아이템을 3.3배 많이 얻고, 2.3배 먼 거리를 이동했으며, 기술 단계(milestone)를 최대 15.3배 빨리 달성했습니다. 스킬 검색의 top-1 정확도는 80.2%였습니다. 마인크래프트 환경의 결과이므로 다른 업무에서도 같은 재사용 효과가 난다고 보지 않습니다. (`[학술]` Wang et al., [arXiv:2305.16291](https://arxiv.org/abs/2305.16291))

**AWM**

AWM(Agent Workflow Memory)은 과거 실행 궤적에서 다시 쓸 수 있는 작업 절차(workflow)를 뽑아 메모리로 남기는 연구입니다. 다음 작업에서는 후보 workflow를 골라 제공하고, 에이전트는 그 절차를 현재 웹사이트와 작업에 맞춰 씁니다. 웹 탐색 벤치마크 Mind2Web과 WebArena에서 기준 방법보다 성공률이 상대적으로 24.6%, 51.1% 높았고, 작업·웹사이트·도메인을 바꾼 평가에서도 8.9점에서 14.0점 높았습니다. 웹 탐색 환경에서 확인한 결과이므로 웹 밖의 작업에도 같은 폭으로 적용된다고 보지 않습니다. (`[학술]` Wang et al., [arXiv:2409.07429](https://arxiv.org/abs/2409.07429))

**컴파운드 엔지니어링**

Every의 개발자 Kieran Klaassen은 이런 흐름을 컴파운드 엔지니어링(compound engineering)이라고 불렀습니다. (`[벤더]` Every, [How I use Claude Code to ship like a team of five](https://every.to/source-code/how-i-use-claude-code-to-ship-like-a-team-of-five-6f23f136-52ab-455f-a997-101c071613aa)) 한 번의 실행 결과를 다음 실행의 입력으로 남기고, 검색·조합·검토를 거치며 작업 능력을 쌓는다는 뜻입니다. 저장한 내용이 오래되거나 틀렸다면 다음 실행에도 같은 오류가 넘어갈 수 있습니다.

**WikiSkill**

WikiSkill은 Google Research가 2026년에 발표한 연구로, 에이전트의 경험을 위키로 쌓아 스킬을 개선합니다. 세 층을 나눠 관리합니다.

- 원문 궤적: 판단의 근거를 확인합니다.
- 위키: 여러 실행에서 얻은 지식을 쌓습니다.
- 스킬: 실제로 실행할 절차를 둡니다.

다음 실행에서는 세 층을 구분해 제공하고, 위키에 쌓인 지식으로 스킬을 갱신합니다. 논문의 Figure 1 캡션은 WikiSkill이 스킬 없는 조건과 기존 스킬 개선 방법(EvoSkill, SkillOpt)보다 일관되게 높다고 설명하고, 초록은 큰 모델일수록 대체로 이점이 크다고 보고합니다. Table 1에서 다섯 벤치마크의 평균 정확도는 Qwen-3.5-4B가 26.2에서 38.5로, Qwen-3.6-27B가 39.4에서 63.3으로 올랐습니다(스킬 없음에서 WikiSkill). 한 모델로 만든 스킬을 다른 모델 패밀리에 옮겨 써도 효과가 있었다고 본문에서 보고합니다. ### 남기는 형태 네 가지

| 남기는 형태 | 다음 실행에서 하는 일 | 장점과 위험 | 먼저 확인할 질문 |
| --- | --- | --- | --- |
| 원문 궤적 | 당시 관찰과 행동의 근거를 다시 확인합니다. | 검토하기 좋지만 깁니다. | 어느 지점까지 믿을 수 있는가 |
| 요약 | 긴 기록에서 현재 판단에 필요한 맥락을 압축합니다. | 빠르지만 내용이 빠질 수 있습니다. | 중요한 실패 조건이 빠지지 않았는가 |
| insight | 여러 실행에서 반복된 판단을 검색해 적용합니다. | 검색하기 좋지만 틀린 내용이 섞일 수 있습니다. | 반례와 갱신 날짜가 있는가 |
| 코드 스킬 | 검증한 절차를 실행 가능한 단위로 재사용합니다. | 바로 실행할 수 있지만 오래된 절차를 그대로 반복할 수 있습니다. | 입력 조건과 되돌리는 방법이 있는가 |

실제 도구에서 이 구분을 보려면, 저장한 세션을 다시 찾아 현재 판단에 넣는 경로부터 확인합니다.

### Hermes에서 확인할 항목

[Hermes](https://github.com/NousResearch/hermes-agent)는 NousResearch가 만든 오픈소스 에이전트(MIT 라이선스)로, 대화 기록을 검색할 수 있는 형태로 남겨 다음 실행에서 다시 찾게 합니다. 아래 내용은 저장소 README에서 확인한 구조이며 성능 근거는 아닙니다. (`[공식 문서]` NousResearch, [hermes-agent README](https://github.com/NousResearch/hermes-agent))

| 확인 항목 | Hermes에서 보는 내용 |
| --- | --- |
| 과거 검색 | FTS5(SQLite 데이터베이스의 전문 검색 기능)로 과거 세션을 검색합니다. |
| 진입점 | 여러 메시지 채널의 요청을 gateway 하나로 받습니다. |
| 다음 컨텍스트 | 세션 이력, memory·user model, 활성 skills를 연결합니다. |
| 확인 과제 | 어떤 기록이 어떤 조건에서 다시 들어오는지 설치본의 설정과 실행 기록으로 확인합니다. |

Hermes에서 볼 것은 기록을 오래 보관한다는 점보다, 과거 세션을 찾아 현재 컨텍스트에 다시 넣는 경로입니다. 검색 결과가 현재 작업과 맞는지, 오래된 설정이 다시 실행되지 않는지, 기억과 스킬이 어떤 권한으로 작동하는지는 따로 점검해야 합니다.

### Claude Code의 스킬

Claude Code의 스킬도 같은 기준으로 봅니다. 스킬(skill)은 `SKILL.md` 파일을 중심으로 지식, 절차, 템플릿, 스크립트를 묶은 디렉터리입니다. 필요한 내용만 단계적으로 불러오도록 만들 수 있고, Claude Code는 개인 또는 프로젝트의 스킬 디렉터리에서 스킬을 찾습니다. (`[공식 문서]` Anthropic, [Claude Code 스킬 문서](https://code.claude.com/docs/en/skills)) 스킬 파일을 추가했다고 경험 축적이 끝나지는 않습니다. 어느 경험에서 왔는지, 언제 검토했는지, 어떤 작업에 적용할지를 함께 남겨야 합니다.

### 메모리 오염(memory poisoning) 실험

ExpeL처럼 모델이 insight를 고르고 지우는 방식에서도 틀린 insight가 남을 수 있다는 점은 경험 기능에 대한 핵심 반론입니다. 그럴듯하지만 틀린 insight 하나가 검색을 통해 여러 작업에 반복해 들어갈 수 있고, 실행 스킬은 그 판단을 더 빨리 행동으로 옮길 수 있습니다. 그래서 운영에서는 스킬을 얼마나 많이 모았는지보다, 틀린 지식을 찾아 고치고 되돌릴 수 있는지가 중요합니다.

이 문제는 작은 주입 실험으로 확인할 수 있습니다.

| 구분 | 내용 |
|---|---|
| 고정 | 작업, 모델, 예산 |
| 조건 | 검증한 insight 또는 스킬을 넣은 조건과, 일부러 틀린 insight 또는 스킬을 넣은 조건 |
| 하나씩 바꿀 정책 | 출처 기록, 만료 기한, 반대 증거, 되돌리기 |
| 기록 | 성공률, 토큰, 오염 전파(틀린 내용이 다른 작업으로 퍼진 정도) |

### 미해결 질문: 남기는 형태만 바꾸면 무엇이 달라지는가

하나의 실행 궤적을 원문, 요약, insight, 코드 스킬 네 형태로 각각 준비합니다. 같은 모델과 같은 작업에서 어느 형태가 다음 행동을 더 잘 돕는지, 토큰을 얼마나 쓰는지, 틀린 내용이 얼마나 퍼지는지 비교합니다. 이 비교는 어떤 형태가 항상 가장 좋은지 정하려는 실험이 아니라, 내 작업에서 재사용하기 쉬운 형태와 검토하기 쉬운 형태가 어디서 갈리는지 찾는 실험입니다.

## 성찰: 실패 기록에서 다음 행동 고치기

성찰은 실패한 실행 기록에서 다음 시도에 바꿀 행동 하나와 그 근거를 짧게 남기는 기능입니다. 답을 길게 평가하는 글이 아니라 다음 행동을 바꾸는 짧은 피드백이어야 합니다. 결과가 틀렸다는 문장만 길게 반복하면 다음 도구 호출과 검색 범위는 바뀌지 않습니다.

### 성찰로 실패 기록 압축하기

**Reflexion**

Reflexion은 실패한 뒤 스스로 성찰 문장을 남겨 다음 시도에 쓰는 방식을 제안한 연구입니다. 모델 가중치를 바꾸지 않고 언어 피드백만으로 시도, 오류, 성찰, 기억으로 이어지는 루프(trial-error-reflection-memory)를 만들었습니다. HotpotQA에서 이전 시도 기록(episodic memory)만 넣은 조건에 성찰을 더하자 정확도가 8%p 올랐습니다. 장기 기억에는 최근 성찰 1~3개만 남겨 다음 시도에 다시 넣었습니다. 모든 실패 기록을 보관하지 않고, 다음 행동을 고칠 근거만 줄여서 다시 넣는 설계입니다. (`[학술]` Shinn et al., NeurIPS 2023, [arXiv:2303.11366](https://arxiv.org/abs/2303.11366))

**Generative Agents**

Generative Agents는 스탠퍼드 연구진이 만든 가상 마을 시뮬레이션 연구로, 기록, 성찰, 검색, 다시 불러오기를 하나의 흐름으로 설계했습니다. 관찰을 memory stream(시간순 기억 목록)에 쌓고, 시간별 성찰로 의미를 묶고, 최근성·관련성·중요도로 찾아 계획에 다시 넣습니다. 이 연구는 작업 성공률이 아니라 에이전트 행동이 그럴듯한지를 평가했으므로, 코딩 하네스의 성능 근거로 넓혀 쓰지 않습니다. (`[학술]` Park et al., UIST 2023, [arXiv:2304.03442](https://arxiv.org/abs/2304.03442))

### 틀린 성찰도 기억이 될 수 있음

성찰이 항상 성능을 높이지는 않았습니다. Reflexion 저자들이 밝힌 결과는 다음과 같습니다.

- MBPP(Python 코딩 벤치마크): Reflexion 77.1%로 GPT-4 기준선 80.1%보다 낮았습니다.
- WebShop: 개선이 없었습니다.

실패 원인을 잘못 설명한 문장이 다음 시도에 다시 들어오면 틀린 교훈이 쌓일 수 있습니다. 그래서 성찰에는 정답처럼 보이는 결론보다 검증할 수 있는 행동을 남깁니다.

| 결론만 남긴 성찰 | 검증할 행동을 남긴 성찰 |
|---|---|
| 검색이 부족했다. | 다음 시도에서는 오류가 난 파일의 호출 지점을 먼저 읽고, 그 뒤에만 검색 범위를 넓힌다. |

다음 실행의 결과가 이 설명을 뒷받침하지 않으면 성찰을 고치거나 버립니다.

### Hermes에서 성찰이 남는 곳

Hermes README에는 성찰 기능이 따로 설명되어 있지 않습니다. 대신 경험·스킬 축적 절에서 본 세션 이력, memory, skills에 성찰 문장도 남을 수 있습니다.

- **남는 상태:** FTS5로 찾은 과거 세션, memory, user model, 활성 스킬이 지속 상태로 남습니다.
- **주요 동작:** 과거 세션을 검색해 다음 요청의 컨텍스트로 다시 가져옵니다. 실패 기록만 골라 가져오는지는 README에서 확인하지 못했습니다.
- **권한 경계:** README에 있는 command approval, DM pairing, container isolation의 실제 적용 범위는 설정과 실행 환경에서 확인해야 합니다.
- **1차 자료:** [hermes-agent README](https://github.com/NousResearch/hermes-agent)와 공개 설정, 실행 기록을 대조합니다.
- **한계:** 성찰이 자동으로 정확해지거나 잘못된 기록이 자동으로 버려진다는 근거는 확인하지 못했습니다.

그래서 성찰 문장에는 작성 시각, 근거가 된 실행 기록, 다음에 바꿀 행동을 함께 연결합니다. 그래야 검색으로 다시 들어온 문장이 어느 실패에서 나왔는지 확인하고, 틀렸을 때 지울 수 있습니다.

### 미해결 질문: 성찰을 몇 개까지 남길까

Reflexion의 최근 1~3개 제한을 그대로 답으로 삼지 않고, 같은 모델·작업·도구 예산에서 성찰을 1개, 3개, 전체로 남긴 조건을 비교합니다. 성공률뿐 아니라 틀린 설명이 다음 시도에 재사용된 횟수와 성찰을 읽는 데 쓴 토큰도 함께 기록합니다.

## 미니멀: 도구와 프롬프트를 줄인 하네스

미니멀은 기능을 적게 넣는 것 자체가 목표가 아니라, 작업을 끝낼 수 있는 가장 작은 루프의 조건을 찾는 기능입니다. 기능 하나를 빼면 컨텍스트 비용과 설정 부담은 줄 수 있지만, 필요한 행동이나 안전 범위까지 빠질 수 있습니다.

### 작은 루프로 작업을 끝낸 사례

| 도구 | 구성 | 주장 | 근거 |
|---|---|---|---|
| mini-swe-agent | 100줄 규모. Bash만 쓰고 셸 상태와 선형 이력을 단순하게 다룹니다. | SWE-bench Verified 74% 초과 | `[벤더]` README 자체 보고 |
| pi | read, write, edit, bash 네 도구. 시스템 프롬프트와 도구 정의를 합쳐 1,000토큰 미만 | 모델이 좋아질수록 복잡한 하네스가 덜 필요할 수 있습니다. | `[벤더]` 저자 에세이 |

**mini-swe-agent**

[mini-swe-agent](https://github.com/SWE-agent/mini-swe-agent)는 SWE-agent 팀이 만든 실험용 최소 에이전트입니다. 스스로를 "100줄 에이전트"라고 소개하며, Bash 말고는 도구를 쓰지 않고 셸 상태와 선형 이력(대화를 순서대로 쌓기만 하는 기록)을 단순하게 다룹니다. README는 SWE-bench Verified에서 74%가 넘는 해결률을 보고하지만, 어떤 모델로 언제 측정했는지는 적혀 있지 않고 독립 논문도 확인하지 못했습니다. (`[벤더]` SWE-agent, [mini-swe-agent README](https://github.com/SWE-agent/mini-swe-agent)) 그래서 이 수치는 풍성한 인터페이스보다 항상 낫다는 근거가 아니라, 복잡한 기능을 빼도 작업을 끝낼 수 있다는 사례로 읽습니다.

**pi**

도구를 줄이는 다른 설계로 pi가 있습니다. [pi](https://github.com/earendil-works/pi)는 개발자 Mario Zechner가 혼자 만든 미니멀 코딩 에이전트입니다. read, write, edit, bash 네 도구와, 합쳐서 1,000토큰이 안 되는 시스템 프롬프트와 도구 정의만으로 설계했습니다. 저자는 모델이 좋아질수록 복잡한 하네스가 덜 필요할 수 있다고 썼습니다. 이는 저자의 설계 가설이며, 이를 뒷받침하는 독립 학술 근거는 확인하지 못했습니다. (`[벤더]` Mario Zechner, [저자 에세이](https://mariozechner.at/posts/2025-11-30-pi-coding-agent/), 2025-11-30)

### pi 비교 카드

| 카드 칸 | pi에서 확인할 내용 |
| --- | --- |
| 진입점 | 터미널에서 여는 대화형 코딩 CLI입니다. |
| 상태 | 세션 안의 대화와 도구 결과가 다음 판단으로 이어집니다. |
| 주요 동작 | read, write, edit, bash 네 도구로 파일을 읽고 바꾸고 명령을 실행합니다. |
| 권한 경계 | 권한 확인 기능(permission system)이 내장되어 있지 않으므로, 컨테이너 같은 외부 격리 환경에 권한 범위를 맡깁니다. |
| 1차 자료 | 공식 저장소와 저자 에세이를 함께 확인합니다. |
| 한계 | Terminal-Bench 2.0을 작업당 5회 실행했다는 언급은 있으나 측정 수치는 확인하지 못했습니다. |

### SWE-agent와 mini-swe-agent의 차이

행동 포맷 절에서 본 SWE-agent는 여기서 풍성한 ACI의 예로 다시 봅니다. SWE-agent의 ACI와 mini-swe-agent의 100줄 Bash 루프는 기능 수만 다른 것이 아닙니다.

- SWE-agent: 모델이 쓸 행동과 관찰을 인터페이스로 정교하게 만듭니다.
- mini-swe-agent: 그 가운데 필요한 부분이 현재 모델에서는 줄어들 수 있는지 묻습니다.

같은 모델과 작업으로 고정하지 않으면 두 결과의 차이를 하네스의 효과라고 말할 수 없습니다.

### "미니멀이면 충분하다"는 아직 가설

mini-swe-agent의 수치와 pi의 주장은 모두 만든 쪽의 자체 보고입니다. 다음 작업에서는 작은 루프가 오히려 실패할 수 있습니다.

- 복잡한 ACI가 필요한 작업
- 긴 이력에서 관찰을 골라야 하는 작업
- 외부 격리로 해결되지 않는 권한 요구가 있는 작업

상용 도구에서도 기능을 덜어 낸 사례가 있습니다.

- Anthropic은 Claude Code의 시스템 프롬프트에서 80% 이상을 지웠는데도 코딩 평가에서 측정 가능한 손실이 없었다고 보고했습니다. 이전 모델의 최악 상황을 막으려고 넣은 규칙이 모델이 나아진 뒤에는 필요 없어졌고, 서로 충돌하는 지시가 지나친 제약이 되었다는 설명입니다. (`[벤더]` Thariq Shihipar, [Claude 블로그](https://claude.com/blog/the-new-rules-of-context-engineering-for-claude-5-generation-models))
- Anthropic의 장기 실행 애플리케이션 개발용 멀티에이전트 하네스에서는 모델이 나아진 것을 확인한 뒤 sprint(작업을 짧은 단위로 나눠 진행하는 구조)를 없애고, 평가를 실행 끝에 한 번만 하도록 줄였습니다. (`[벤더]` Prithvi Rajasekaran, [Anthropic Engineering](https://www.anthropic.com/engineering/harness-design-long-running-apps))

두 사례 모두 현재 모델이 스스로 해내는 일을 시험해 보고 불필요한 장치를 덜어 낸 경우입니다.

### 미해결 질문: 모델이 좋아지면 격차가 줄어드는가

모델 세대가 오를수록 풍성한 ACI와 Bash만 쓰는 선형 이력 루프의 성능 격차가 줄어들까요. mini-swe-agent의 74% 초과가 나온 모델·시점·작업 조건을 먼저 고정한 뒤, 같은 조건에서 두 루프의 성공률, 토큰, 실행 횟수, 권한 요청을 비교합니다. 조건을 고정하기 전에는 숫자를 재현했다고 말하지 않습니다.

## 권한·보안: 허용 범위와 격리

파일을 읽는 능력, 명령을 실행하는 능력, 외부 메시지를 받는 능력은 서로 다른 위험을 만듭니다. 권한·보안 기능은 이 세 행동의 진입점과 실행 범위를 나눠, 입력 하나가 곧바로 모든 도구 권한으로 이어지지 않게 합니다.

### 차단 수치와 작업 결과를 함께 보는 연구

외부 문서에 숨긴 지시가 모델의 행동을 바꾸는 공격을 프롬프트 인젝션(prompt injection)이라고 합니다. 아래 연구들은 이 공격을 얼마나 막는지와 함께, 막는 동안 작업을 얼마나 끝내는지를 다룹니다.

| 연구 | 한 일 | 결과 (해당 조건) |
|---|---|---|
| Spotlighting | 외부 입력의 출처를 표시하는 변환을 적용 | 공격 성공률 50% 초과에서 2% 미만으로 하락 |
| ToolEmu | 고위험 도구 사용을 에뮬레이션으로 평가 (도구 묶음 36개, 사례 144개) | 가장 안전한 에이전트도 23.9% 실패 |
| CaMeL | 권한, 데이터 흐름, 제어 흐름을 구조적으로 분리 | 설계 제안 |
| AgentDojo | 공격 성공률과 작업 완수율을 함께 측정 (작업 97개, 보안 테스트 629개) | 벤치마크 제안 |

**Spotlighting**은 외부 입력에 출처를 표시하는 변환을 넣어 모델이 지시와 데이터를 구분하게 하는 방법입니다. 이 변환을 적용했을 때 공격 성공률이 50% 초과에서 2% 미만으로 내려갔다고 보고했습니다. 출처를 구분하면 공격을 줄일 수 있다는 근거이지만, 저자 실험 조건의 결과이므로 모든 문서와 공격에 같은 방어 효과가 난다고 보지 않습니다. (`[학술]` Hines et al., [arXiv:2403.14720](https://arxiv.org/abs/2403.14720))

**ToolEmu**는 위험한 도구 사용을 실제로 실행하지 않고 언어 모델로 흉내 내어(에뮬레이션) 안전성을 평가하는 방법입니다. 고위험 도구 묶음(toolkit) 36개와 사례 144개에서, 가장 안전한 에이전트도 23.9%의 사례에서 실패했습니다. 에뮬레이션으로 얻은 결과이므로 실제 컨테이너 격리나 다른 업무 환경의 실패율로 넓혀 읽지 않습니다. (`[학술]` Ruan et al., ICLR 2024, [arXiv:2309.15817](https://arxiv.org/abs/2309.15817))

**CaMeL**은 권한, 데이터 흐름, 제어 흐름을 구조적으로 나눠 외부 입력이 민감한 도구를 제어하지 못하게 하는 설계를 제안한 연구입니다. 민감한 데이터를 다루는 도구와 외부 입력이 만나는 곳에서, 다음 행동을 무엇이 정하는지를 따로 관리하자는 접근입니다. (`[학술]` Debenedetti et al., [arXiv:2503.18813](https://arxiv.org/abs/2503.18813))

**AgentDojo**는 에이전트의 공격 성공률과 작업 완수율을 함께 측정하는 벤치마크입니다. 많이 막으면 안전해 보이지만 작업도 끝내지 못할 수 있기 때문에 두 값을 함께 봅니다. (`[학술]` Debenedetti et al., NeurIPS 2024 Datasets and Benchmarks, [arXiv:2406.13352](https://arxiv.org/abs/2406.13352))

외부 메시지가 들어오는 환경에서는 그 메시지를 어느 세션과 어떤 권한으로 연결하는지까지 확인해야 합니다. [OpenClaw](https://docs.openclaw.ai/)는 Discord, Slack, Telegram 같은 여러 채널에서 상시 대기하는 오픈소스 에이전트입니다. 이 질문을 OpenClaw의 설정으로 살펴봅니다.

### OpenClaw 비교 카드 (권한 경계)

| 카드 칸 | OpenClaw에서 확인할 내용 |
| --- | --- |
| 진입점 | Gateway가 Discord, Slack, Telegram을 비롯한 여러 채널의 메시지를 받습니다. |
| 상태 | Gateway가 채널 연결과 세션 라우팅을 관리하며, 에이전트·작업 공간·발신자별로 세션을 분리합니다. |
| 주요 동작 | 상시 대기하는 Gateway가 들어온 메시지를 해당 세션과 에이전트로 라우팅합니다. |
| 권한 경계 | pairing, `allowFrom`, 그룹 멘션 요구(`requireMention`)로 메시지를 보낼 수 있는 범위를 설정합니다. |
| 1차 자료 | OpenClaw [공식 문서](https://docs.openclaw.ai/)에서 구조와 설정을 확인합니다. |
| 한계 | 이 권한 설계의 공격 방어나 작업 성능을 검증한 학술 근거는 확인하지 못했습니다. |

메시지를 보낼 수 있는 범위는 세 설정으로 정합니다.

| 설정 | 하는 일 | 정하지 않으면 |
|---|---|---|
| pairing | 채널과 Gateway가 처음 연결될 때 어느 발신자와 페어링할지 정합니다. 페어링을 승인하면 그 발신자의 메시지가 정해진 채널을 거쳐 해당 세션으로 라우팅됩니다. | 기본 정책에서는 페어링되지 않은 발신자를 승인 절차로 분리합니다. 이 정책을 바꾸거나 승인 없이 페어링을 허용하면, 처음 보는 발신자가 신뢰할 수 있는 진입점인지 확인할 수 없습니다. |
| `allowFrom` | 페어링된 채널 안에서도 메시지를 보낼 수 있는 발신자 범위를 정합니다. 목록 밖 발신자의 메시지는 에이전트 실행으로 이어지지 않습니다. | 채널의 접근 규칙만으로는 허용할 발신자가 명시적으로 좁혀지지 않습니다. |
| 그룹 멘션 요구 | 그룹 대화에서 에이전트를 멘션한 메시지만 처리하고, 멘션 없는 메시지는 무시합니다. | 일반 대화가 Gateway가 처리할 요청과 섞일 수 있습니다. |

세 설정은 메시지를 보낸 사람과 에이전트 호출 여부를 가리는 설정입니다. 도구 실행이 허용된 뒤 피해 범위를 가두는 격리 환경과는 다른 단계에서 작동합니다.

### Claude Code 권한 모드와 opencode 권한 규칙

**Claude Code 권한 모드**는 어떤 도구 행동을 자동으로 허용하고 어떤 행동을 승인 대상으로 둘지 정합니다. 모드에 따라 파일 변경이나 명령 실행이 바로 실행되거나 승인 화면에서 멈춥니다. 모드를 따로 정하지 않으면 기본 승인 흐름을 따르므로, 어떤 행동이 왜 승인 없이 실행됐는지는 설정에서 따로 확인해야 합니다. (`[공식 문서]` Anthropic, [Claude Code 권한 문서](https://code.claude.com/docs/en/permissions))

**opencode 권한 규칙**은 도구와 행동별로 `allow`(허용), `ask`(확인), `deny`(거부)를 정합니다. 같은 행동에 규칙이 여러 개 맞으면 마지막에 맞은 규칙이 적용되고, 에이전트별 규칙이 전역 규칙보다 우선합니다. 그래서 특정 에이전트의 편집은 확인을 받고 다른 에이전트의 읽기는 허용하는 식으로 동작을 나눌 수 있습니다. 규칙을 정하지 않으면 전역 규칙과 기본 정책에 따라 판단되므로, 작업별 예외가 어느 규칙에서 왔는지 읽기 어렵습니다. (`[공식 문서]` opencode, [권한 문서](https://opencode.ai/docs/permissions))

두 문서는 구조를 설명할 뿐이며 정량 실험 근거는 아닙니다.

### 권한과 격리

권한은 누가 무엇을 허용할지를 다루고, 격리는 허용된 실행의 피해를 어디에 가둘지를 다룹니다. 하나가 다른 하나를 대신하지는 않습니다.

| 구분 | 권한 | 격리 |
|---|---|---|
| 다루는 것 | 누가 무엇을 허용하는가 | 허용된 실행의 피해를 어디에 가두는가 |
| 작동 시점 | 실행 전 판단 | 실행된 프로세스가 닿는 파일·네트워크·운영체제 범위 |
| 예 | Claude Code 권한 규칙, opencode `allow`·`ask`·`deny` | Docker 컨테이너 같은 외부 격리 환경, Claude Code의 Bash 샌드박스 |

도구마다 두 가지를 배치한 방식이 다릅니다.

- pi는 내장 권한 체계가 없어 실행한 사용자의 권한으로 도구를 실행하며, 격리가 필요하면 Docker 같은 외부 환경에 맡깁니다. (`[벤더]` Mario Zechner, [저자 에세이](https://mariozechner.at/posts/2025-11-30-pi-coding-agent/))
- Claude Code는 권한 규칙을 도구 안에 두어 행동의 허용과 승인을 먼저 다룹니다. 여기에 더해 운영체제가 셸 명령이 닿는 파일과 네트워크를 제한하는 Bash 샌드박스도 제공합니다(macOS, Linux, WSL2). (`[공식 문서]` Anthropic, [샌드박스 문서](https://code.claude.com/docs/en/sandboxing))
- OpenClaw는 에이전트·작업 공간·발신자별로 세션을 나눠 대화 상태가 섞이는 범위를 줄입니다. 이 세션 분리는 컨테이너 격리와 같지 않으며, host에서 실행하는 도구의 범위는 별도의 sandbox 정책과 실행 환경에서 확인해야 합니다.

pi와 Claude Code는 같은 위험을 없애는 하나의 해법이 아니라, 권한 판단과 실행 환경을 서로 다르게 배치한 사례입니다. macOS `sandbox-exec`나 Docker의 구체적인 명령은 이 노드에서 다루지 않습니다.

### 비교 실험: 허용 규칙 하나 바꾸기

권한 설정은 안전하다는 선언으로 끝나지 않습니다. 같은 작업에서 허용 규칙을 하나 바꾸고 다음 값을 함께 기록합니다.

- 처리된 요청 수
- 승인 횟수
- 막힌 정상 작업
- 의도하지 않은 도구 호출

외부 메시지를 받는 도구라면 허용되지 않은 발신자의 메시지와 멘션 없는 메시지가 실제로 어느 세션에도 들어가지 않는지도 확인합니다. 권한을 촘촘히 할수록 공격을 덜 받는다고 바로 결론 내리지 않습니다. 정상 작업을 막은 비율과 어떤 설정이 그 차단을 만들었는지를 함께 남겨야 다음 설정을 바꿀 근거가 생깁니다.

## 평가 루프: 결과와 실행 궤적 나눠 확인하기

최종 패치가 맞았다는 사실만으로는 검색을 낭비했는지, 우연히 성공했는지, 권한을 과하게 썼는지 구분할 수 없습니다. 평가 루프는 최종 패치가 작업을 해결했는지와, 그 결과에 이른 검색·편집·실행 과정이 적절했는지를 나눠 확인합니다. 에이전트가 거친 검색·편집·실행을 순서대로 남긴 기록을 실행 궤적(trajectory)이라고 하고, 이후에는 궤적이라고 줄여 씁니다.

### 구성요소와 결과를 따로 측정하기

**SWE-agent의 ablation**

SWE-agent의 ablation은 검색, 편집, 관찰 형식처럼 인터페이스 구성요소를 바꾸고 결과를 비교할 수 있음을 보였습니다(행동 포맷 절 참고). 같은 모델이라도 무엇을 읽고 어떤 형식으로 행동했는지를 고정해야 구성요소 하나의 기여를 물을 수 있습니다. 이 결과는 SWE-bench Lite와 논문에 적힌 실행 조건의 결과이므로, 다른 모델이나 저장소에서도 같은 구성요소가 같은 기여를 한다고 보지 않습니다. (`[학술]` Yang et al., NeurIPS 2024, [arXiv:2405.15793](https://arxiv.org/abs/2405.15793))

**SWE-bench**

[SWE-bench](https://www.swebench.com/)는 실제 GitHub 이슈와 저장소로 코딩 에이전트를 측정하는 벤치마크입니다. 12개 Python 저장소에서 모은 문제 2,294개로 이루어지고, 최종 패치를 두 종류의 테스트로 채점합니다. (`[학술]` Jimenez et al., ICLR 2024 Oral, [arXiv:2310.06770](https://arxiv.org/abs/2310.06770))

| 테스트 | 확인하는 것 |
|---|---|
| FAIL_TO_PASS | 고치기 전에 실패하던 테스트가 패치 뒤에 통과하는지(이슈 해결) |
| PASS_TO_PASS | 원래 통과하던 테스트가 패치 뒤에도 계속 통과하는지(기존 기능 유지) |

문제를 푸는 루프와 채점하는 루프를 나눈 구조입니다. 다만 테스트 통과가 유지보수 품질 전체와 같지는 않고, 공개된 문제라서 모델의 학습 데이터에 섞였을 위험(contamination)도 남습니다.

**SWE-bench Verified**

SWE-bench Verified는 OpenAI가 SWE-bench 저자들과 함께 2024년 8월에 공개한 500개 문제 모음입니다. Python 경험이 있는 개발자 93명이 SWE-bench 문제를 검토해, 문제 설명과 테스트로 풀 수 있다고 판단한 문제만 남겼습니다. (`[공식 문서]` OpenAI, [Introducing SWE-bench Verified](https://openai.com/index/introducing-swe-bench-verified/)) 이 검토는 문제가 풀 수 있는지에 대한 근거이며, 특정 하네스가 더 낫다는 근거는 아닙니다.

### SWE-agent에서 평가에 쓰는 기록

| 기능 | 확인할 기록 | 1차 자료 |
| --- | --- | --- |
| SWE-agent 평가 루프 | 문제 설명·저장소, 궤적, 패치, 테스트 결과를 같은 실행 단위에 남겨 결과와 경로를 함께 검토합니다. | SWE-agent 공개 저장소와 실행 기록, ACI 논문을 확인합니다. |

궤적은 모델의 의도나 패치가 맞는지를 스스로 증명하지 않습니다. 대신 어느 검색이 먼저 있었는지, 테스트 실패 뒤 무엇을 편집했는지, 어떤 명령이 실행됐는지를 1차 자료로 남깁니다. 이 기록으로 다음을 다시 확인합니다.

- 다음 행동을 바꾼 관찰이었는가
- 같은 탐색을 반복했는가
- 승인한 권한이 과도했는가

### 통과 결과만으로 부족한 이유

정적 벤치마크의 테스트 통과는 장기 유지보수, 설명 가능성, 코드 검토, 의존성 변화 대응을 재지 못합니다. 공개된 문제와 해답이 학습 데이터나 도구 설정에 섞였을 가능성도 비교를 흐릴 수 있습니다. 그러므로 결과 점수와 함께 실행 환경, 모델, 권한 정책, 비용, 승인 횟수, 궤적을 남기고, 실제 코드 검토는 따로 합니다. 어떤 궤적 신호로 좋은 하네스를 가려낼 수 있는지는 아직 남은 질문입니다.

## 일곱 도구 비교표로 후보 고르기

표에서 빈칸이 적은 도구가 아니라, 내 병목 칸에 근거가 있는 도구를 고릅니다.

먼저 내 작업에서 확인할 병목을 하나 정하고, 표에서는 그 병목에 해당하는 열을 읽습니다. 도구 이름을 한꺼번에 외우려는 표가 아닙니다. 행동 포맷부터 미니멀, 권한까지 각 기능에서 도구가 무엇을 하는지 보고, 마지막 평가 열에서 무엇이 확인됐고 무엇이 남았는지 확인합니다.

Claude Code처럼 소스가 공개되지 않은 도구는 내부 구현을 확인할 수 없고, 공개 코드가 있는 도구도 설치 버전, 로컬 설정, 실행 기록 없이 안전하다고 증명할 수 없습니다. 그래서 이 표를 제품의 우열이나 안전 판정으로 읽지 않습니다. 표는 앞 절에서 읽은 카드와 다시 확인할 근거를 한자리에 모은 것입니다.

### 내 병목으로 후보를 두 개로 줄이기

표를 다 읽은 뒤에도 모든 열을 같은 무게로 비교하지 않습니다. 첫 절의 세 질문으로 후보를 고릅니다.

1. 내 작업의 병목 기능을 하나 적습니다.
2. 그 기능을 확인할 근거가 카드의 1차 자료에 있는지 확인합니다.
3. 확인하지 못한 범위가 현재 작업에서 허용되는지 판단합니다.

예를 들어 컨텍스트가 병목이면 세션, 관찰 요약, repo map 가운데 내 작업에서 볼 수 있는 기록을 가진 도구를 남깁니다. 권한이 병목이면 문서의 기능 설명만 보고 남기지 않고, 현재 설정과 승인 기록을 확인할 수 있는 도구를 남깁니다. 필요한 범위가 확인되지 않았다면 탈락이라고 단정하지 않고 다음 확인 대상으로 적습니다. 이 과정을 거쳐 후보를 두 개로 줄입니다.

비교표는 후보를 줄여 줄 뿐입니다. 최종 판정에는 자기 환경에서 만든 실행 기록이 필요합니다.

## 과제 - 도구의 불편을 개선 가설로 바꿔 실험하기

앞 절들에서 제안한 비교 실험과 같은 방식으로, 지금 쓰는 하네스 도구의 불편 하나를 골라 실험으로 확인합니다. 기준 도구는 Claude Code, opencode, Antigravity 등 실제로 사용하고 있는 도구 가운데에서 고릅니다. 논문의 수치는 정답이 아니라 조건이 붙은 가설이므로, 자기 환경에서 재현하거나 반박할 근거로만 씁니다.

### 1단계. 지금 쓰는 도구의 불편 떠올리기

최근 작업에서 반복해서 멈추거나 되돌린 장면 하나를 고릅니다. 승인 요청이 너무 자주 나왔는지, 필요한 파일을 놓쳤는지, 기록이 길어져 중요한 조건을 잃었는지, 결과를 검토하기 어려웠는지를 행동과 결과로 적습니다. "도구가 불편합니다"처럼 넓게 적지 않고, 어떤 작업에서 어떤 입력 뒤에 무엇이 늦어지거나 잘못되었는지를 적습니다.

이 불편이 정말 하네스 기능에서 온 것인지도 함께 의심합니다. 모델 선택, 작업 난이도, 저장소 상태, 네트워크, 자신의 요청 문장이 원인일 수도 있습니다. 뒤의 실험은 불편을 확인하는 데 그치지 않고, 원인이 하네스 기능이었다는 설명까지 검토합니다.

### 2단계. 불편을 개선 가설과 근거로 바꾸기

불편을 줄일 수 있는 하네스 기능 하나를 골라 반증 가능한 개선 가설로 씁니다. 반증 가능하다는 것은 결과가 반대로 나오면 가설이 틀렸다고 말할 수 있다는 뜻입니다. 예를 들어 승인 대기로 작업이 멈췄다는 관찰은 "권한 규칙의 범위를 조정하면 완료 시간이 줄어든다"는 가설로 바꿀 수 있습니다.

가설은 1단계에서 적은 불편에서 시작합니다. 행동 포맷 절부터 평가 루프 절까지 인용한 연구에서는 그 가설을 뒷받침하거나 반박할 근거를 찾습니다. 비슷한 조건의 연구에서 원문 문장, 측정 수치, 연구가 검증하지 않은 것, 근거 종류를 옮겨 적습니다. 근거 종류 표기 `[학술]`, `[벤더]`, `[공식 문서]`(첫 절 참고)를 숨기지 않으며, 벤더의 자체 보고는 독립 비교가 아니라는 점도 보고서에 남깁니다.

### 3단계. 한 항목만 바꾸는 실험 설계하기

| 구분 | 내용 |
|---|---|
| 고정 | 모델, 작업 묶음, 토큰 또는 시간 예산 |
| 조작 (하나만) | 관찰 개수 N(컨텍스트 관리 절에서 본 최근 관찰 개수), 편집 포맷, 권한 정책, 메모리 표현 가운데 하나. 사용하는 도구에서 실제로 바꿀 수 있는 항목만 고릅니다. 예를 들어 Claude Code와 opencode에서는 권한 규칙과 메모리 표현(`notes/` 같은 기록 파일)을 바꿀 수 있고, 편집 포맷은 Aider의 `--edit-format`처럼 포맷을 고를 수 있는 도구에서만 바꿀 수 있습니다. |
| 이중 측정 | 해결률과 함께 토큰, 승인 횟수, 시간을 측정 |
| 반론 | 예상 반론 하나 이상과, 그 반론과 가설을 구분할 통제 조건 |
| 사전 판정식 | 결과를 보기 전에, 어떤 수치면 가설을 지지하고 어떤 수치면 버릴지 정한 기준 |

고정 조건, 조작 항목, 이중 측정, 사전 판정식이 분리되어야 불편의 원인을 다른 조건과 구별할 수 있습니다. 실험 설계 용어로는 고정은 통제 변인, 조작은 조작 변인, 측정은 종속 변인에 해당합니다.

### 4단계. 결과를 GitHub 보고서로 제출하기

실험을 실행한 뒤 GitHub 저장소에 Markdown 보고서를 올립니다. 저장소는 다음 구조로 만듭니다.

```
my-harness-experiment/
├── README.md          # 가설, 실험 설계, 결과, 다시 검토할 조건 요약
├── experiment-01.md   # 실험 보고서
├── traces/            # 수정 전후의 실행 기록
└── fixture/           # 고정한 작업 입력 원본
```

`README.md`에는 아래 "보고서에 담을 내용" 가운데 가설, 실험 설계 표, 판정 결과, 재검토 조건을 요약하고, 전체 내용은 `experiment-01.md`에 씁니다. 실행했다면 수정 전후(before·after)의 실행 기록(trace)을 `traces/`에 넣거나 링크합니다. 실행하지 못했다면 실행하지 못한 이유와 사전 판정식을 남깁니다.

보고서에는 불편의 실제 원인이 선택한 하네스 기능이었는지 돌아보는 문단을 넣습니다. 가설과 다른 결과가 나왔다면 실패로 감추지 않고, 무엇이 원인 설명을 약하게 만들었는지와 다음 실험에서 고정하거나 바꿀 조건을 적습니다.

보고서를 모두 올렸으면 저장소 URL을 LMS에 제출합니다. LMS에서는 제출한 뒤에 수정할 수 없습니다.

### 평가 기준

제출한 보고서는 다음 기준으로 봅니다.

- 가설이 반증 가능한가
- 고정, 조작, 측정이 분리되었는가
- 근거 종류를 구분해 인용했는가
- 불편의 원인이 하네스 기능이었는지 돌아보았는가

### 과제 진행 예시

Claude Code에서 컴파운드 엔지니어링, 즉 한 번 얻은 경험을 다음 실행에 쌓아 성능을 올리는 접근이 실제로 효과가 있는지 확인해 봅니다. 경험·스킬 축적 절에서 본 ExpeL, Voyager, WikiSkill과 같은 방향의 주장입니다.

**가설.** 같은 작업을 다시 풀 때, 첫 시도의 기록을 `notes/` 폴더에 남기고 다음 요청에서 그 기록을 참조하게 하면, 기록 없이 시작할 때보다 두 번째 시도가 빠르고 정확해집니다.

**실험 작업 3개.** 세 작업 모두 fixture로 입력을 고정합니다.

1. `parse-log`: fixture 로그 파일 `sample.log`에서 오류만 골라 표로 정리하는 스크립트 `parse.py` 작성 (확인: `test_parse.py`)
2. `refactor-util`: fixture 유틸리티 `util.py`의 함수 3개에 타입 힌트와 독스트링 추가 (확인: `test_util.py`)
3. `fix-failing`: fixture 테스트 `test_me.py`의 실패 2건을 통과하도록 수정 (확인: `test_me.py`)

**사전 준비.** 교안은 fixture 파일을 제공하지 않으므로 직접 만듭니다. 저장소의 `fixture/` 폴더에서 Claude Code를 실행하고 다음과 같이 요청합니다.

```
sample.log, util.py, test_me.py 실습용 fixture 만들어 줘. test_me.py는 실패 2건이 나오게 해 줘.
작업 확인용 테스트도 함께 만들어 줘. test_parse.py는 parse.py가 sample.log의 오류 줄만 골라 표로 출력하는지, test_util.py는 util.py의 함수 3개에 타입 힌트와 독스트링이 있는지 확인하게 해 줘. 두 테스트는 지금은 실패해야 해.
```

한 번 만든 fixture는 내용을 바꾸지 않습니다. 입력이 같아야 조건 사이의 차이가 작업 난이도가 아니라 경험 기록 때문이라고 말할 수 있습니다. 작업이 `util.py`와 `test_me.py`를 고치므로, `fixture/`는 원본으로만 두고 시도할 때마다 복사해서 씁니다. 각 작업은 괄호 안의 테스트가 통과하면 끝난 것으로 봅니다.

**실험 설계.**

| 구분 | 내용 |
|---|---|
| 고정 | Claude Code, 같은 모델, 같은 세 작업, 같은 fixture |
| 조작 | 경험 기록 조건: 첫 시도 뒤 `notes/`에 접근법과 주의점을 남기고 다음 요청에서 참조하게 함. 대조 조건: 기록 없이 새로 시작 |
| 측정 | 두 번째 시도의 소요 시간, 수정 횟수, 테스트 통과까지의 시도 수, 토큰 |
| 사전 판정식 | 경험 기록 조건이 세 작업 가운데 두 개 이상에서 시간과 시도 수가 모두 적으면 가설 지지, 그 외에는 기각 |

측정값은 다음처럼 셉니다.

| 측정값 | 세는 방법 |
|---|---|
| 소요 시간 | 작업을 마친 뒤 `/usage`를 입력하고 Session 블록의 `Total duration (wall)`을 적습니다. |
| 토큰 | 같은 Session 블록의 `Usage by model` 줄에서 input과 output 토큰 수를 적습니다. |
| 수정 횟수 | 화면에 표시된 파일 편집·작성 도구 호출의 수를 셉니다. |
| 테스트 통과까지 시도 수 | 해당 작업의 테스트가 통과할 때까지 학생이 Claude Code에 요청을 보낸 횟수를 셉니다. |

`/usage`의 Session 블록은 `/clear`로 새 세션을 시작하면 0부터 다시 셉니다(v2.1.211 이상). 그보다 낮은 버전에서는 Claude Code를 새로 실행해야 0부터 셉니다. (`[공식 문서]` Anthropic, [Claude Code 비용 문서](https://code.claude.com/docs/en/costs))

**실행 절차.**

1. 저장소 밖에 작업 폴더를 하나 만들고, 그 안에 `fixture/`의 원본을 복사해 폴더 A(경험 기록 조건)와 폴더 B(대조 조건)를 만듭니다. 저장소 밖에 두면 작업 중에 만든 파일과 `notes/`가 저장소의 `fixture/` 원본, 보고서 파일과 섞이지 않습니다.
2. 두 폴더에서 세 작업을 한 번씩 풀고, 작업마다 소요 시간과 시도 수를 기록합니다.
3. 폴더 A에만 `notes/`를 만들고, 작업마다 접근법, 만난 오류, 해결 방법을 세 줄 이내로 적습니다. 폴더 B에는 기록을 남기지 않습니다.
4. 두 폴더의 작업 파일을 `fixture/`의 원본으로 되돌립니다. 폴더 A의 `notes/`는 그대로 둡니다.
5. 두 폴더에서 세 작업을 다시 풀게 하고 같은 지표를 기록합니다. 폴더 A에서는 요청할 때 `notes/`의 파일을 함께 참조하라고 지시합니다.
6. 두 조건의 기록을 나란히 놓고 사전 판정식에 넣습니다.

시도할 때마다 Claude Code를 새로 실행하거나 `/clear`로 대화를 비운 뒤 시작합니다. 앞 시도의 대화가 남아 있으면 두 조건의 차이가 `notes/`에서 왔다고 말할 수 없습니다.

**기록 표.** 두 번째 시도의 결과를 아래 표에 채웁니다. 화면에서 확인하지 못한 값은 지어내지 말고 "미확인"으로 적습니다.

| 작업 | 조건 | 소요 시간 | 수정 횟수 | 테스트 통과까지 시도 수 | 토큰 |
|---|---|---|---|---|---|
| parse-log | 경험 기록 | | | | |
| parse-log | 대조 | | | | |
| refactor-util | 경험 기록 | | | | |
| refactor-util | 대조 | | | | |
| fix-failing | 경험 기록 | | | | |
| fix-failing | 대조 | | | | |

**결과 보고.** 세 작업의 두 조건 기록을 표로 남기고, 판정 결과(지지 또는 기각)를 함께 적습니다. 기각되어도 실험이 실패한 것은 아닙니다. 어느 작업에서 기록이 도움이 되지 않았는지가 다음 실험을 설계하는 근거가 됩니다.

**제출.** 보고서는 GitHub 저장소의 `experiment-01.md`에 아래 "보고서에 담을 내용" 순서대로 씁니다.

자기 도구의 불편에서 출발할 때도 작업을 fixture로 고정하고, 조작을 한 가지로 좁히고, 결과가 나오기 전에 판정식을 쓰는 흐름은 같습니다.
